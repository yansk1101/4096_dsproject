# **Exploratory Data Analysis (EDA)**

**Basic information of the DataFrame**
1. Consists 1253 rows
2. Float for meterological variables
3. Integer for H1, H3, and AandB while float for B 
4. Binary value for pandemic_flag
5. Float for OxCGRT health index

**Statistical information for columns**

1. Variance is much larger than the mean for influenza case counts 
    -  Shows overdispersion (massive amount of a specific value in the corresponding column)
    -  Need to use Quasi-Poisson / Negative Binomial / Tweedie as the distribution family for DLNM to handle overdispersion
2. `ContainmentHealthIndex_Average` ranges from 0 to 82.14

**Missing value identification**
1. `B` exist one missing value
2. All other columns do not have missing values

**Ratio of counts in influenza case counts**

As the statistical information shows the variance of influenza cases is much larger than the mean, we want to find out what causes the overdispersion


**Correlations**

Correlation matrix is constructed and visualized using a heatmap

1. Correlation between different meterological values
2. Correlation between meterological variables and influenza values
3. Correlation between meterological variables and influenza counts with lag

**Variance Inflation Factor (VIF) test**

used to check multicollinearity - the degree of high correlation between meterological variables

 

**Line graph**

used to see the overall trend of influenza case counts in terms of types

**Scatterplot**

used to identify the relationship between  meteorlogical data and influenza case counts for different types of influenza

The x-axis are the meterological data, while the y-axis represents different type of influenza

From the figure, we can claim that our previous finding of overdispersion is correct by seeing that the majority of data points lie on the bottom of the graph with some points lying on the top of the graph. This finding points out the appropriate probability distribution of family that we should use for advanced modelling. 

All relationships are non-linear and heteroscedasticity, showing that non-linear models should be used instead of linear models.
