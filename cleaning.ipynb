```python
path = os.getcwd()
path += "\\dataset\\IVRS_data"
csv_files = glob.glob(os.path.join(path, "*.csv"))

df_list = []
for f in csv_files:
    year = int(re.findall(r"\d+", os.path.basename(f))[0])

    if year in [i for i in range(2014, 2020)]:
        skip_no = 5
    elif year in [i for i in range(2020, 2025)]:
        skip_no = 6
    else: 
        skip_no = 8

    temp_df = pd.read_csv(f, skiprows=1, skipfooter=skip_no, header=[0, 1, 2], engine="python")     
    
    if year in [i for i in range(2023, 2025)]:
        temp_df = temp_df.iloc[:, :15]
        temp_df.columns = [
            "Week", "Date",
            "No. of specimen tested", 
            "Type A positive", "Type A positive %", "Type A Subtyping performed", 
            "Type A subtype H1 - No.", "Type A subtype H3 - No.", "Type A subtype H5 - No.", "Type A subtype H7 - No.", "Type A subtype H9 - No.", 
            "Type B - No.", "Type B - %", 
            "Type C - No.", "Type C - %"
        ]
        temp_df["Year"] = year
   
    elif year in [2025, 2026]:
        temp_df = temp_df.iloc[:, :16]
        temp_df.columns = [
            "Week", "Date",
            "No. of specimen tested", 
            "Type A positive", "Type A positive %", "Type A Subtyping performed", 
            "Type A subtype H1 - No.", "Type A subtype H3 - No.", "Type A subtype H5 - No.", "Type A subtype H7 - No.", "Type A subtype H9 - No.", 
            "Type B - No.", "Type B - %", "Type B - Lineage deetermination performed", 
            "Type C - No.", "Type C - %"
        ]
        temp_df["Year"] = year
       
    else:
        temp_df = temp_df.iloc[:, :17]
        temp_df.columns = [
            "Week", "Date", 
            "No. of specimen tested", 
            "Type A subtype H1 - No.", "Type A subtype H1 - %", 
            "Type A subtype H3 - No.", "Type A subtype H3 - %", 
            "Type A subtype H5 - No.", "Type A subtype H5 - %", 
            "Type A subtype H7 - No.", "Type A subtype H7 - %", 
            "Type A subtype H9 - No.", "Type A subtype H9 - %", 
            "Type B - No.", "Type B - %", 
            "Type C - No.", "Type C - %"
        ]
        temp_df["Year"] = year
        
    df_list.append(temp_df)

ivrs_df = pd.concat(df_list, ignore_index=True)

ivrs_df
```

```python
# Function for reading weather datas
def read_file():

    csv_files = glob.glob(os.path.join(path, "*.csv"))
    
    # Create an empty list for storing the DataFrames
    df_list = []

    # Read all the files and process it 
    for f in csv_files:
        # Retrieve the second row to get the location (station) English name
        temp_df = pd.read_csv(f, skiprows=1, header=None, nrows=1, usecols=[0], names=["raw_line"])
        location = temp_df["raw_line"].str.split("- ")[0][1] # manually corrected the second line in the dataset for HKO
        ''' 
        if "(" in location:
            location = location.split(" (")[0]
        '''
        # Read the remaining file and add a new column 'Location' to it 
        temp_df = pd.read_csv(f, skiprows=2, skipfooter=3, engine="python")
        temp_df["Location"] = location
        # Append the DataFrae into the list 
        df_list.append(temp_df)

    # Concat all the DataFrames into one 
    df = pd.concat(df_list, ignore_index=True)
    return df
    
```

```python
# Function for data type conversion
def data_type_conversion(df_name):
    df_name["Year"] = df_name["Year"].astype(int)
    df_name["Month"] = df_name["Month"].astype(int)
    df_name["Day"] = df_name["Day"].astype(int)
    df_name["Date"] = df_name["Year"].astype(str) + "/" + df_name["Month"].astype(str) + "/" + df_name["Day"].astype(str)
    df_name.drop(["Year", "Month", "Day"], axis=1, inplace=True)
    
    drop_index = df_name[df_name["Date"] == "1900/2/29"].index
    df_name.drop(drop_index, inplace=True)
    df_name["Date"] = pd.to_datetime(df_name["Date"], format="mixed")
    
```

```python
# Apply the function to the DataFrames
for df in df_list:
    if df.equals(RH_df):
        continue
    else:
        data_type_conversion(df)


drop_index = RH_df[RH_df["Month"].isna()].index
RH_df.drop(drop_index, inplace=True)

data_type_conversion(RH_df)
```

```python
# Function for filtering the rows for the date
def drop_rows(df_name):
    drop_index = df_name[((df_name["Date"] > pd.to_datetime("2026/07/26")) | (df_name["Date"] < pd.to_datetime("2013/01/01")))].index
    df_name.drop(drop_index, inplace=True)
```

```python
# Apply the function to the DataFrames
for df in df_list:
    drop_rows(df)
```

```python
weather_df = max_temp_df.drop("data Completeness", axis=1).merge(min_temp_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(mean_temp_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(RH_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(SR_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(rainfall_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(AP_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(dewpt_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(WS_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])
weather_df = weather_df.merge(UV_df.drop("data Completeness", axis=1), how="outer", on=["Date", "Location"])

weather_df = weather_df.reindex(["Date", "Max temperature", "Min temperature", "Mean temperature", "Rainfall", "Mean humidity", "Rainfall", "Air pressure", "Solar radiation", "Dew point temperature", "Wind speed", "Mean UV"], axis=1)

display(weather_df)
weather_df.info()
```

```python
# change datatype to float of the numeric columns
# to-do, no rainfall
weather_df[["Max temperature", "Min temperature", "Mean temperature", "Mean humidity", "Solar radiation", "Air pressure", "Dew point temperature", "Wind speed", "Mean UV"]] = weather_df[["Max temperature", "Min temperature", "Mean temperature", "Mean humidity", "Solar radiation", "Air pressure", "Dew point temperature", "Wind speed", "Mean UV"]].astype(float)

weather_df.info()
```

```python
# group the rows according to the date
# to-do, no rainfall
cleaned_weather_df = weather_df.drop(["Rainfall"], axis=1).groupby(pd.Grouper(key="Date", axis=0, freq="W-SUN", closed="left", label="left")).mean().round(2)
display(cleaned_weather_df)
```

```python
# Add the columns 'Year', 'Week', and 'From'
cleaned_weather_df["Year"] = pd.Series(cleaned_weather_df.index, index=cleaned_weather_df.index).apply(lambda d: Week.fromdate(d).year)
cleaned_weather_df["Week"] = pd.Series(cleaned_weather_df.index, index=cleaned_weather_df.index).apply(lambda d: Week.fromdate(d).week)
cleaned_weather_df["From"] = cleaned_weather_df.index

cleaned_weather_df["Diurnal temperature range"] = cleaned_weather_df["Max temperature"] - cleaned_weather_df["Min temperature"]

# Re-index
cleaned_weather_df.index = pd.RangeIndex(0, cleaned_weather_df.shape[0])
cleaned_weather_df = cleaned_weather_df.reindex(["Year", "Week", "From", "Diurnal temperature range", "Mean temperature", "Mean humidity", "Solar radiation", "Air pressure", "Dew point temperature", "Wind speed", "Mean UV"], axis=1)
cleaned_weather_df.columns = ["Year", "Week", "From", "Diurnal temperature range (°C)", "Mean temperature (°C)", "Mean humidity (%)", "Solar radiation (MJ/m²)", "Air pressure (hPa)", "Dew point temperature (°C)", "Wind speed (km/h)", "Mean UV"]

# Filter the DataFrame by date
cleaned_weather_df = cleaned_weather_df[cleaned_weather_df["From"] > pd.to_datetime("1996-12-28")]

display(cleaned_weather_df)
cleaned_weather_df.info()
```
