```python
import glob
import os
import re
import pandas as pd
import numpy as np
from epiweeks import Week
```

```python
def correct_file_with_epiweeks(df, form):
    df['From_Date'] = pd.to_datetime(df['From'], format=form)

    df["Year"] = df["From_Date"].apply(lambda date: Week.fromdate(date).year)
    df["Week"] = df["From_Date"].apply(lambda date: Week.fromdate(date).week)
    
    df = df.drop(columns=['From_Date'])

    return df
```

```python
# Function for reading weather datas
def read_file():

    csv_files = glob.glob(os.path.join(path, "*.csv"))
    
    # Create an empty list for storing the DataFrames
    df_list = []

    # Read all the files and process it 
    for f in csv_files:
        if re.search(r"_HKO|_SHA|_LFS|_TKL|_TUN|_HKS|_WGL|_JKB|_CCH|_KP|_EPC|_KAT|_PLC|_SLW|_SKG|_TAP|_TYW|_SEK|_TPO|_CPH|_HKA|_SF|_SHL|_NP|_SE|_GI|_BHD", f):
            # Retrieve the second row to get the location (station) English name
            temp_df = pd.read_csv(f, skiprows=1, header=None, nrows=1, usecols=[0], names=["raw_line"])
            location = temp_df["raw_line"].str.split("- ")[0][1] # manually corrected the second line in the dataset for HKO
            ''' 
            if "(" in location:
                location = location.split(" (")[0]
            '''
            # Read the remaining file and add a new column 'Location' to it 
            temp_df = pd.read_csv(f, skiprows=2, skipfooter=3, engine="python")
            temp_df["Location"] = location
            # Append the DataFrae into the list 
            df_list.append(temp_df)

    # Concat all the DataFrames into one 
    df = pd.concat(df_list, ignore_index=True)
    return df
    
```

```python
# Function for data type conversion
def data_type_conversion(df_name):
    df_name["Year"] = df_name["Year"].astype(int)
    df_name["Month"] = df_name["Month"].astype(int)
    df_name["Day"] = df_name["Day"].astype(int)
    df_name["Date"] = df_name["Year"].astype(str) + "/" + df_name["Month"].astype(str) + "/" + df_name["Day"].astype(str)
    df_name.drop(["Year", "Month", "Day"], axis=1, inplace=True)
    
    drop_index = df_name[df_name["Date"] == "1900/2/29"].index
    df_name.drop(drop_index, inplace=True)
    df_name["Date"] = pd.to_datetime(df_name["Date"], format="mixed")
    
```

```python
# Apply the function to the DataFrames
for df in df_list:
    if df.equals(RH_df):
        continue
    else:
        data_type_conversion(df)


drop_index = RH_df[RH_df["Month"].isna()].index
RH_df.drop(drop_index, inplace=True)

data_type_conversion(RH_df)
```

```python
# Function for filtering the rows for the date
def drop_rows(df_name):
    drop_index = df_name[((df_name["Date"] > pd.to_datetime("2026/09/01")) | (df_name["Date"] < pd.to_datetime("2002/09/01")))].index
    df_name.drop(drop_index, inplace=True)
```

```python
# Apply the function to the DataFrames
for df in df_list:
    drop_rows(df)
```

```python
# group the rows according to the date
cleaned_weather_df = weather_df.groupby(pd.Grouper(key="Date", axis=0, freq="W-SUN", closed="left", label="left")).agg({
    "Mean temperature": "median", 
    "Mean humidity": "median", 
    "Air pressure": "median", 
    "Solar radiation": "median", 
    "Dew point temperature": "median", 
    "Wind speed": "median", 
    "Mean UV": "median",
    "Max temperature": "max",
    "Min temperature": "min",
}).round(2)
display(cleaned_weather_df)
```

```python
# Add the columns 'Year', 'Week', and 'From'
cleaned_weather_df["Year"] = pd.Series(cleaned_weather_df.index, index=cleaned_weather_df.index).apply(lambda d: Week.fromdate(d).year)
cleaned_weather_df["Week"] = pd.Series(cleaned_weather_df.index, index=cleaned_weather_df.index).apply(lambda d: Week.fromdate(d).week)
cleaned_weather_df["From"] = cleaned_weather_df.index

cleaned_weather_df["Diurnal temperature range"] = cleaned_weather_df["Max temperature"] - cleaned_weather_df["Min temperature"]

# Re-index
cleaned_weather_df.index = pd.RangeIndex(0, cleaned_weather_df.shape[0])
cleaned_weather_df = cleaned_weather_df.reindex(["Year", "Week", "From", "Diurnal temperature range", "Mean temperature", "Mean humidity", "Solar radiation", "Air pressure", "Dew point temperature", "Wind speed", "Mean UV"], axis=1)
cleaned_weather_df.columns = ["Year", "Week", "From", "Diurnal temperature range (°C)", "Mean temperature (°C)", "Mean humidity (%)", "Solar radiation (MJ/m²)", "Air pressure (hPa)", "Dew point temperature (°C)", "Wind speed (km/h)", "Mean UV"]

# Filter the DataFrame by date
cleaned_weather_df = cleaned_weather_df[cleaned_weather_df["From"] >= pd.to_datetime("2002-09-01")]

display(cleaned_weather_df)
cleaned_weather_df.info()
```

```python
path = os.getcwd()
path += "\\dataset\\Population.csv"
df_pop = pd.read_csv(path)

df_pop = df_pop.iloc[[3, 4, 6]]
df_pop.drop(columns=["Unnamed: 1"], inplace=True)
df_pop = df_pop.T
df_pop.index = range(df_pop.shape[0])
df_pop.columns = ["Year", "Reference time-point", "Population"]
df_pop.drop([0], inplace=True)
display(df_pop.head())
df_pop.info()
df_pop["Year"] = pd.to_numeric(df_pop["Year"])
```

```python
df["Reference time-point"] = np.where(df["Week"] >= 23, "Year-end", "Mid-year")

df = pd.merge(df, df_pop, on=["Year", "Reference time-point"], how="left")

df.loc[df[df["Population"].isna()].index, "Population"] = df_pop.loc[df_pop[df_pop["Year"]==2026].index, "Population"].values[0]
df["Population"] = pd.to_numeric(df["Population"])
df["Log_population"] = np.log(df["Population"])

df.drop(columns="Reference time-point", inplace=True)
```
