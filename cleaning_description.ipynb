# **Data Preprocessing**

**Regarding influenza dataset**
1. use `correct_file_with_epiweeks()` to get the correct MMWR date format year and week column
2. Filter out unwanted columns and rows
3. Data type conversion 
4. Concatenate samples retreived from the two sources

**Regarding meterological dataset**
1. use `read_file()` to read datasets for each meterological factors in the selected stations and add a column 'Location' for further processing 
2. Filter out unwanted columns and rows
3. Rename columns
4. Drop rows with missing values (keep rows with incomplete data label)
5. Concatenate all individual meterological DataFrame into one single DataFrame recording all meterological variables along the date and location colnums
6. Turn daily records into weekly records
    - Use `max`, `min` for maximum and minimum temperature
    - Use `median` for others as large sample size with extreme values, using median could be less sensitive for extreme values

**Regrading the combined dataset**
1. Spot out missing values, and input the values manually with the values obtained from Flu Express
2. Correct invalid data
3. Change the index values (0 to 1253)


**Handling of pandemic period**
1. `pandemic_flag`: indicate whether the specific week is in pandemic period or not
2. `ContainmentHealthIndex_Average`: indicate how pandemic affects influenza transmission 
    - obtained from OxCGRT Github
    - answers "How" instead of Yes or No
    - a scale from 0 to 100
https://www.bsg.ox.ac.uk/research/covid-19-government-response-tracker
