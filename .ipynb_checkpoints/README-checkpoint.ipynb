### 'df' Data Dictionary:
| Column | Description | Value |
| --- | --- | --- |
|`Year`|Year (MMWR format)|integer (2013 to 2026)|
|`Week`|Week number (MMWR format)|integer (1 to 53)|
|`From`|Week Start Date (MMWR format)|datetime|
|`Diurnal temperature range (°C)`|Temperature range|numeric (float)|
|`Mean temperature (°C)`|Mean temperature|numeric (float)|
|`Mean humidity (%)`|Mean relative humidity|numeric (float)|
|`Solar Radiation (MJ/m²)`|Global solar radiation|numeric (float)|
|`Air pressure (hPa)`|Air pressure|numeric (float)|
|`Dew point temperature (°C)`||numeric (float)|
|`mean UV`||numeric (float)|
|`H1`|Type A subtype H1(N1) influenza detections|numeric (float)|
|`H3`|Type A subtype H3(N2) influenza detections|numeric (float)|
|`B`|Type B influenza detections|numeric (float)|
|`AandB`|Type A and B influenza detections|numeric (float)|
|`H1_proportion`|NA from 1997 to 2013, H1:number of specimen tested|
|`H3_proportion`|NA from 1997 to 2013, H3:number of specimen tested|
|`B_proportion`|NA from 1997 to 2013, B:number of specimen tested|
|`AandB_proportion`|NA from 1997 to 2013, (A+B):number of specimen tested|
