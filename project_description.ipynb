# **Project description (Flu & Meteorlogical triggers)**

**Research Question**

How strongly do seasonal weather fluctuations (temperature drops, relative humidity levels, solar radiation) correlate with weekly influenza surveillance positivity rates and outpatient consultation surges? 

**Guiding Questions**
1. Quantifies the difference in the severity or incidence of influenza during winter compared to any summer outbreaks. 
2. Determine if there’s a delayed effect between specific weather conditions and a rise in flu cases. 
3. International Flu Strain & Seasonal Sync Benchmarking: Compares Hong Kong’s flu season patterns and dominant strains with international trends. 
4. Can Machine Learning Model predicts the influenza outbreaks in October 2026 accurately with the historical data?
